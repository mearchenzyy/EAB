### Construct dictionary storing sorted filenames
construct a dictionary where the keys are data file types (like 'CB_FTXX45_YY_d=0') and the values are a list of data filenames in order (['line_001.txt','line_008.txt',...]).

```python
def check_for_missing_files(num_of_lines,l):
    missing_files=[]
    for i in range (1,num_of_lines+1):
        if i not in l:
            missing_files.append(i)
    if len(missing_files) !=0:
        print ('missing files:')
        print (missing_files)
    return missing_files 
```

```python
def sort_data_files(dict_keys,data_filenames,l):
    dict_keys_len=len(dict_keys)
    sorted_datafiles={}
    for key in dict_keys:
        sorted_datafiles[key]=[]
    for idx,filename in enumerate(data_filenames):
        sorted_datafiles[dict_keys[l[idx]%dict_keys_len-1]].append(filename)
    return sorted_datafiles
```

```python
def copy_txt_file(source_path, destination_folder):
    """Copies a .txt file to a specified folder. Probably want to rename the file--change line number) 

    Args:
        source_path (str): The path to the .txt file to be copied.
        destination_folder (str): The path to the folder where the file should be copied.
    """
    try:
        # Ensure the destination folder exists. Make dir if doesn't exsit
        os.makedirs(destination_folder, exist_ok=True)
        # Construct the full destination path
        destination_path = os.path.join(destination_folder, os.path.basename(source_path))
        # Copy the file
        shutil.copy(source_path, destination_path)
        print(f"File '{source_path}' copied to '{destination_path}' successfully.")
    except FileNotFoundError:
        print(f"Error: File '{source_path}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
        
# Example usage:
source_file = '/Users/yzhu/yzhu_work/gates projects/EAB/data/01092025/'
destination_directory = "path/to/your/destination/folder"
copy_txt_file(source_file, destination_directory)
```

```python
def copy_txt_file(root_folder, source_folder_name, sorted_filenames):
    """Copies a .txt file to a specified folder. Probably want to rename the file--change line number) 

    Args:
        source_path (str): The path to the .txt file to be copied.
        destination_folder (str): The path to the folder where the file should be copied.
    """
    source_path=root_folder+source_folder_name
    for subfolder,filenames in sorted_filenames.items():
        path= root_folder+'sorted_data/'+subfolder+'/'
        try:
            # Ensure the destination folder exists. Make dir if doesn't exsit
            os.makedirs(path, exist_ok=True)
        except Exception as e:
            print(f"An error occurred: {e}")
            # Construct the full destination path
            # destination_path = os.path.join(destination_folder, os.path.basename(source_path))
            # Copy the file
        for file in filenames:
            try:
                shutil.copy(source_path+file, path)
                print(f"File '{file}' copied to '{path}' successfully.")
            except FileNotFoundError:
                print(f"Error: File '{file}' not found.")
            except Exception as e:
                print(f"An error occurred: {e}")
            
```
